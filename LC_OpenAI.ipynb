# Agents
```
Agents combine language models with tools to create systems that can reason about tasks, decide which tools to use, and iteratively work towards solutions.
```
```
1.   Params
    *   model
    *   tools
    *   middleware
    *   system prompt
    *   content schema
    *   name
    *   checkpoint
    *   response format
```

````python
import os, requests
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    url = f"https://api.weatherapi.com/v1/current.json?key=8acb7523f2c34e9d83391219262205&q={city}"

    response = requests.get(url)

    return response.json()["current"]["temp_c"]

model = ChatOpenAI(
    model="gpt-5.4",
    temperature=0.2,
    max_tokens=300,
    timeout=30
)

agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

response = agent.invoke({
    "messages" : {
        "role":"user",
        "content": "What is the temperatur in Delhi"
    }
})

print(response['messages'][-1].content)
````

# Messages

````python
from langchain.chat_models import init_chat_model
from langchain.messages import SystemMessage, HumanMessage, AIMessage
````

````python
model = init_chat_model('gpt-5.4')
sys_msg = SystemMessage("You are a helpful coding assistant.")
usr_msg = HumanMessage("Who is the Coach of Indian cricket team")
messages = [sys_msg, usr_msg]
res = model.invoke(messages)
````
