## Async Generator
```
1. asyncio.run(main())
2. asyncio.gather(func1(), func2())
    - run together
    - wait for all
3. asyncio.create_task(func())
    - run parallely
```

````python
# This cell code workd in the online compiler

import asyncio
async def counter():
    count=0
    while True:
        yield count
        count+=1

async def main():
    async for i in counter():
        print(i)
        if i == 5:
            break

# asyncio.run(main())
````

````python
# Battery percentage decreasing

import asyncio

async def battery():
    percent = 100
    while True:
        yield percent
        percent-=1
        await asyncio.sleep(1)

async def main():
    async for per in battery():
        if per == 90:
            break
        print(per)

asyncio.run(main())
````

````python
# Temperature increasing

import asyncio

async def temperature():
    temp = 25
    while True:
        yield temp
        temp+=1
        await asyncio.sleep(0.5)

async def main():
    async for t in temperature():
        print(t)
        if t == 30:
            break

asyncio.run(main())
````
